```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif, RFE
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
X = data[['pH', 'TDS', 'Turbidity', 'Temp']]
y = data['Recommended_Crop']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
selector = SelectKBest(score_func=f_classif, k=3)
X_train_kbest = selector.fit_transform(X_train, y_train)
X_test_kbest = selector.transform(X_test)

print("Selected Features (KBest):", selector.get_support())
```

```python
rfe_model = RandomForestClassifier()
rfe = RFE(rfe_model, n_features_to_select=3)
X_train_rfe = rfe.fit_transform(X_train, y_train)
X_test_rfe = rfe.transform(X_test)

print("Selected Features (RFE):", rfe.support_)
```

```python
models = {
    "Logistic Regression": LogisticRegression(),
    "KNN": KNeighborsClassifier(n_neighbors=3),
    "SVM": SVC(),
    "Decision Tree": DecisionTreeClassifier(),
    "Random Forest": RandomForestClassifier()
}
```

```python
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    
    print("\nModel:", name)
    print("Accuracy:", accuracy_score(y_test, y_pred))
    print(classification_report(y_test, y_pred))
```

```python
for name, model in models.items():
    scores = cross_val_score(model, X, y, cv=5)
    print(f"{name} Average CV Accuracy: {scores.mean():.4f}")
```

```python
for name, model in models.items():
    scores = cross_val_score(model, X, y, cv=5)
    print(f"{name} Average CV Accuracy: {scores.mean():.4f}")
```

```python
rf = RandomForestClassifier()
rf.fit(X, y)

importance = rf.feature_importances_

feature_importance = pd.DataFrame({
    'Feature': ['pH', 'TDS', 'Turbidity', 'Temp'],
    'Importance': importance
}).sort_values(by='Importance', ascending=False)

print(feature_importance)
```

```python
accuracies = {}
best_model = None
best_score = 0
best_model_name = ""
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    accuracies[name] = acc

    if acc > best_score:
        best_score = acc
        best_model = model
        best_model_name = name

print("Best Model:", best_model_name)
print("Best Accuracy:", best_score)
```

```python
plt.figure()
plt.bar(accuracies.keys(), accuracies.values())
plt.xlabel("Models")
plt.ylabel("Accuracy")
plt.title("Model Accuracy Comparison")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import joblib
import pandas as pd

# 1. Load your models
model = joblib.load('best_model.pkl')
scaler = joblib.load('scaler.pkl')

# 2. Define the Live Demo Function
def run_live_demo(ph, tds, temp, turb, season, knn_status="Normal"):
    # --- LOGIC STEP 1: Filter out bad water first ---
    if knn_status == "Abnormal" or tds > 2000:
        return "No Crops Suitable (Water Quality Issues)"
        
    # --- LOGIC STEP 2: Prepare EXACTLY 4 features for the model ---
    # The model expects only these 4 columns in this exact order
    input_features = pd.DataFrame([[ph, tds, turb, temp]], 
                                  columns=['pH', 'TDS', 'Turbidity', 'Temp'])
    
    # --- LOGIC STEP 3: Scale & Predict ---
    # Scale the 4 features
    input_scaled = scaler.transform(input_features)
    
    # Predict the crop using only the 4 scaled features
    predicted_crop = model.predict(input_scaled)[0]
    
    return predicted_crop

# 3. RUN THE LIVE TEST CASES
print("--------------------------------------------------")
print(" LIVE DEMO: CROP & SEASON RECOMMENDATION FEATURE")
print("--------------------------------------------------\n")

# Case 1 (Perfect Kharif)
res_1 = run_live_demo(ph=6.5, tds=450, temp=27, turb=5, season='Kharif')
print(f"Test Case 1 (Kharif) -> Inputs: pH 6.5, TDS 450 | Expected: Groundnut")
print(f"Model Output: {res_1}\n")

# Case 2 (Perfect Summer)
res_2 = run_live_demo(ph=7.2, tds=1300, temp=32, turb=8, season='Summer')
print(f"Test Case 2 (Summer) -> Inputs: pH 7.2, TDS 1300 | Expected: Pomegranate")
print(f"Model Output: {res_2}\n")

# Case 3 (High TDS Fail)
res_3 = run_live_demo(ph=7.0, tds=2800, temp=25, turb=5, season='Any')
print(f"Test Case 3 (High TDS Fail) -> Inputs: pH 7.0, TDS 2800 | Expected: No Crops Suitable")
print(f"Model Output: {res_3}\n")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, accuracy_score
import joblib

# 1. Load your data and tools
df = pd.read_csv('anantapur_balanced_crops.csv')
model = joblib.load('best_model.pkl')
scaler = joblib.load('scaler.pkl')

# 2. Isolate the 4 sensors and the actual answers
X = df[['pH', 'TDS', 'Turbidity', 'Temp']]
y_actual = df['Recommended_Crop']

# 3. Ask the AI to predict the whole dataset
X_scaled = scaler.transform(X)
y_pred = model.predict(X_scaled)

# 4. Print the overall score
accuracy = accuracy_score(y_actual, y_pred)
print(f"🏆 OVERALL AI ACCURACY: {accuracy * 100:.2f}%\n")

# 5. Build the Confusion Matrix Visual
cm = confusion_matrix(y_actual, y_pred, labels=model.classes_)

plt.figure(figsize=(12, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=model.classes_, yticklabels=model.classes_)
plt.title('AI Accuracy: Expected Crop vs. Predicted Crop', fontsize=16, fontweight='bold')
plt.xlabel('What the AI Predicted', fontsize=12, fontweight='bold')
plt.ylabel('What the Crop Actually Was', fontsize=12, fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
