```python
#.................................................prediction(TESTING)............................................#

import os
import time
import joblib
import warnings
import pandas as pd
warnings.filterwarnings(action = 'ignore')
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split


def hydro_model(hydro_value):    
    clf = joblib.load('knn.sav')
    for i in range(len(hydro_value)):
        hydro_value[i]=float(i)
        
    new_pred = clf.predict([hydro_value])
    if new_pred==[0.]:
        result = 'Abnormal'
        print(result)
    else:
        result = 'Normal'
        print(result)
    f = open("output.txt","w")
    f.write(result)
    f.close()
    print('\n')


if __name__ == '__main__':
    TestData="Hydro_Test"
    while True:
        for(direcpath,direcnames,files) in os.walk(TestData):
            for file in files:
                if 'txt' in file:
                    time.sleep(1)
                    filename= TestData +'/'+ file
                    print('Filename:',filename)
                    with open('input.txt', 'r') as myfile:
                        data = myfile.read()
                        testdata = data.split("\n")
                        clf = hydro_model(testdata)
                    os.remove(filename)

```
