```python
import pandas as pd
import matplotlib.pyplot as plt
import re

# Read each full line as text (no splitting)
raw = pd.read_csv("data_log.csv", header=None, dtype=str)

# Combine columns into single line (handles split timestamp)
raw["line"] = raw.apply(lambda r: " ".join([x for x in r.astype(str).tolist() if x != "nan"]), axis=1)

# Extract timestamp + 4 numbers from each line
# Works for formats like:
# "2026-02-07 13:08:13 26 35 450 7.1"
pattern = r"(\d{4}-\d{2}-\d{2}\s+\d{2}:\d{2}:\d{2}).*?([-+]?\d*\.?\d+).*?([-+]?\d*\.?\d+).*?([-+]?\d*\.?\d+).*?([-+]?\d*\.?\d+)"

rows = []
for s in raw["line"]:
    m = re.search(pattern, s)
    if m:
        rows.append(m.groups())

df = pd.DataFrame(rows, columns=["timestamp", "temperature", "turbidity", "tds", "ph"])

# Convert types
df["timestamp"] = pd.to_datetime(df["timestamp"])
for c in ["temperature", "turbidity", "tds", "ph"]:
    df[c] = pd.to_numeric(df[c], errors="coerce")

df = df.dropna()

print(df.head())
print("Cleaned shape:", df.shape)

# Plot
plt.figure(figsize=(10,5))
plt.plot(df["timestamp"], df["ph"], label="pH")
plt.plot(df["timestamp"], df["tds"], label="TDS")
plt.plot(df["timestamp"], df["turbidity"], label="Turbidity")
plt.plot(df["timestamp"], df["temperature"], label="Temperature")

plt.xlabel("Time")
plt.ylabel("Sensor Values")
plt.title("Temporal Variation of Water Quality Parameters")
plt.legend()
plt.grid(True, alpha=0.3)
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig("Figure_Sensor_Trends.png", dpi=300)
plt.show()
```
