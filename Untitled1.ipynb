```python
import pandas as pd
import matplotlib.pyplot as plt
import re

# Read each full line as text (no splitting)
raw = pd.read_csv("data_log.csv", header=None, dtype=str)

# Combine columns into single line (handles split timestamp)
raw["line"] = raw.apply(lambda r: " ".join([x for x in r.astype(str).tolist() if x != "nan"]), axis=1)

# Extract timestamp + 4 numbers from each line
# Works for formats like:
# "2026-02-07 13:08:13 26 35 450 7.1"
pattern = r"(\d{4}-\d{2}-\d{2}\s+\d{2}:\d{2}:\d{2}).*?([-+]?\d*\.?\d+).*?([-+]?\d*\.?\d+).*?([-+]?\d*\.?\d+).*?([-+]?\d*\.?\d+)"

rows = []
for s in raw["line"]:
    m = re.search(pattern, s)
    if m:
        rows.append(m.groups())

df = pd.DataFrame(rows, columns=["timestamp", "temperature", "turbidity", "tds", "ph"])

# Convert types
df["timestamp"] = pd.to_datetime(df["timestamp"])
for c in ["temperature", "turbidity", "tds", "ph"]:
    df[c] = pd.to_numeric(df[c], errors="coerce")

df = df.dropna()

print(df.head())
print("Cleaned shape:", df.shape)

# Plot
plt.figure(figsize=(10,5))
plt.plot(df["timestamp"], df["ph"], label="pH")
plt.plot(df["timestamp"], df["tds"], label="TDS")
plt.plot(df["timestamp"], df["turbidity"], label="Turbidity")
plt.plot(df["timestamp"], df["temperature"], label="Temperature")

plt.xlabel("Time")
plt.ylabel("Sensor Values")
plt.title("Temporal Variation of Water Quality Parameters")
plt.legend()
plt.grid(True, alpha=0.3)
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig("Figure_Sensor_Trends.png", dpi=300)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
import numpy as np

# 1) Read raw (no headers)
raw = pd.read_csv("data_log.csv", header=None)

print("Raw shape:", raw.shape)
print(raw.head())

# 2) Build timestamp from first 2 columns (date + time) if present
#    and keep remaining 4 columns as sensors
#    Expected: [date, time, temp, turbidity, tds, ph]  -> 6 cols
#    BUT you reported 5 cols; commonly it's [datetime, temp, turb, tds, ph]
#    We'll handle both.

if raw.shape[1] >= 6:
    # date + time in col0 & col1
    raw["timestamp"] = pd.to_datetime(raw[0].astype(str) + " " + raw[1].astype(str), errors="coerce")
    sensors = raw.iloc[:, 2:6].copy()
elif raw.shape[1] == 5:
    # Either:
    # col0 is full datetime string OR date only; try parse directly
    raw["timestamp"] = pd.to_datetime(raw[0].astype(str), errors="coerce")
    sensors = raw.iloc[:, 1:5].copy()
else:
    raise ValueError("Unexpected number of columns. Please print raw.head() and raw.shape.")

# 3) Name sensor columns (ASSUMED ORDER)
# Change order here if your dataset order is different!
sensors.columns = ["temperature", "turbidity", "tds", "ph"]

# 4) Convert to numeric and drop bad rows
for c in sensors.columns:
    sensors[c] = pd.to_numeric(sensors[c], errors="coerce")

df = pd.concat([raw["timestamp"], sensors], axis=1).dropna()
df = df.sort_values("timestamp")

print("Cleaned shape:", df.shape)
print(df.head())

# 5) Normalize (0–1) for clean single-graph visualization
scaler = MinMaxScaler()
scaled = scaler.fit_transform(df[["ph", "tds", "turbidity", "temperature"]])
df_scaled = pd.DataFrame(scaled, columns=["ph", "tds", "turbidity", "temperature"])
df_scaled["timestamp"] = df["timestamp"].values

# 6) Plot normalized trends (publication-friendly)
plt.figure(figsize=(10,5))
plt.plot(df_scaled["timestamp"], df_scaled["ph"], label="pH (norm)")
plt.plot(df_scaled["timestamp"], df_scaled["tds"], label="TDS (norm)")
plt.plot(df_scaled["timestamp"], df_scaled["turbidity"], label="Turbidity (norm)")
plt.plot(df_scaled["timestamp"], df_scaled["temperature"], label="Temperature (norm)")

plt.title("Normalized Temporal Variation of Water Quality Parameters")
plt.xlabel("Time")
plt.ylabel("Normalized Value (0–1)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig("Figure_Sensor_Trends_Normalized.png", dpi=300)
plt.show()
```
